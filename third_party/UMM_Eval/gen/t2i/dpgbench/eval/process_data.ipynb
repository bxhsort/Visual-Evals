```python
import json
from tqdm import tqdm
import os
def read_jsonl(path):
    with open(path, 'r', encoding='utf-8') as f:
        return [json.loads(line) for line in f]
def write_jsonl(data, path):
    with open(path, 'w', encoding='utf-8') as f:
        for item in data:
            f.write(json.dumps(item, ensure_ascii=False) + '\n')
```

```python
all_files = list(os.listdir('/mnt/afs/wupenghao/workspace/Neo_base/inference_mot/dpg_bench/prompts'))
all_prompts = []
for file in all_files:
    name = file.replace('.txt', '')
    with open(f"/mnt/afs/wupenghao/workspace/Neo_base/inference_mot/dpg_bench/prompts/{file}", "r", encoding="utf-8") as f:
        text = f.read().strip()
    all_prompts.append({'id':name, 'prompt':text})
len(all_prompts)
```

```python
write_jsonl(all_prompts, "prompts.jsonl")
```
